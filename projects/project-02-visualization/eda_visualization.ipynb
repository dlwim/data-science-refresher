```python
from datetime import datetime; 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib 

print(f"Run: {datetime.now().strftime("%Y-%m-%d %H:%M:%S")}")

print("✅ Environment setup successful!")
print(f"pandas: {pd.__version__}")
print(f"numpy: {np.__version__}")
print(f"matplotlib: {matplotlib.__version__}")
print(f"seaborn: {sns.__version__}")
```

```python
print(f"Run: {datetime.now().strftime("%Y-%m-%d %H:%M:%S")}")

# Load the Titanic dataset
titanic_df = pd.read_csv('../../datasets/titanic.csv')
# Display first few rows
print(f"Dataset shape: {titanic_df.shape[0]} rows, {titanic_df.shape[1]} columns")
print("\nFirst 5 rows:")
titanic_df.head(5)
```

```python
print(f"Run: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")

ages = titanic_df["Age"].dropna()

fig, ax1 = plt.subplots(figsize=(10, 6))

# Primary y-axis (density) — needed for the KDE to share the same scale
ax1.hist(ages, bins=30, density=True, color="steelblue", edgecolor="white",
         alpha=0.8, label="Age distribution")
sns.kdeplot(ages, color="darkblue", linewidth=2, label="KDE", ax=ax1)

median_age = ages.median()
ax1.axvline(median_age, color="crimson", linestyle="--", linewidth=1.5,
            label=f"Median age: {median_age:.1f}")

ax1.set_title("Distribution of Passenger Ages")
ax1.set_xlabel("Age (years)")
ax1.set_ylabel("Density", color="steelblue")
ax1.tick_params(axis="y", labelcolor="steelblue")

# Secondary y-axis (counts) — scaled to match the density axis exactly
ax2 = ax1.twinx()
ax2.set_ylim(ax1.get_ylim()[0] * len(ages), ax1.get_ylim()[1] * len(ages))
ax2.set_ylabel("Passenger count", color="gray")
ax2.tick_params(axis="y", labelcolor="gray")

ax1.legend()
ax1.text(0.97, 0.95,
         f"n = {len(ages)} passengers\n({titanic_df['Age'].isna().sum()} ages missing)",
         transform=ax1.transAxes, ha="right", va="top", fontsize=9, color="gray")

plt.show()
```

```python
print(f"Run: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")

# Fare has very few missing values but extreme outliers — drop nulls only
#fares = titanic_df["Fare"].dropna()
fares = titanic_df[titanic_df["Fare"] > 0]["Fare"].dropna()

fig, ax1 = plt.subplots(figsize=(10, 6))

# Fare is heavily right-skewed (a few 1st class fares exceed £500)
# Using bins=40 to capture detail in the crowded lower range
ax1.hist(fares, bins=40, density=True, color="steelblue", edgecolor="white",
         alpha=0.8, label="Fare distribution")

# KDE overlay — will show the strong right skew clearly
sns.kdeplot(fares, color="darkblue", linewidth=2, label="KDE", ax=ax1)

# Median reference — more meaningful than mean here given the skew
median_fare = fares.median()
ax1.axvline(median_fare, color="crimson", linestyle="--", linewidth=1.5,
            label=f"Median fare: £{median_fare:.2f}")

ax1.set_title("Distribution of Passenger Fares")
ax1.set_xlabel("Fare (£)")
ax1.set_xscale("log")  
ax1.set_ylabel("Density", color="steelblue")
ax1.tick_params(axis="y", labelcolor="steelblue")

# Secondary y-axis showing raw passenger counts
ax2 = ax1.twinx()
ax2.set_ylim(ax1.get_ylim()[0] * len(fares), ax1.get_ylim()[1] * len(fares))
ax2.set_ylabel("Passenger count", color="gray")
ax2.tick_params(axis="y", labelcolor="gray")

ax1.legend()
ax1.text(0.97, 0.95,
         f"n = {len(fares)} passengers\n({titanic_df['Fare'].isna().sum()} fares missing)",
         transform=ax1.transAxes, ha="right", va="top", fontsize=9, color="gray")

plt.show()
```

```python
print(f"Run: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")

# Box plot - Passenger Fares by Class
plt.figure(figsize=(10, 6))
#sns.boxplot(data=titanic_df, x='Pclass', y='Fare', palette='Blues')
sns.boxplot(data=titanic_df, x='Pclass', y='Fare', hue='Pclass', palette='Blues', legend=False)
plt.xlabel('Passenger Class')
plt.ylabel('Fare (£)')
plt.title('Distribution of Passenger Fares by Class')
for pclass in sorted(titanic_df['Pclass'].unique()):
    fares = titanic_df[titanic_df['Pclass'] == pclass]['Fare'].dropna()
    q1, median, q3 = fares.quantile([0.25, 0.5, 0.75])
    iqr = q3 - q1
    x_pos = pclass - 1  # seaborn positions classes at 0, 1, 2

    plt.text(x_pos + 0.1, median + 5, f'Median: £{median:.0f}', va='center', fontsize=8)
    plt.text(x_pos + 0.1, q1 - 5, f'Q1: £{q1:.0f}', va='center', fontsize=8)
    plt.text(x_pos + 0.1, q3 + 5, f'Q3: £{q3:.0f}', va='center', fontsize=8)
    plt.text(x_pos + 0.1, q3 + 15, f'IQR: £{iqr:.0f}', va='center', fontsize=8, color='navy')

plt.ylim(-20, 320)
plt.show()
```

```python
print(f"Run: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")

plt.figure(figsize=(10, 6))
sns.boxplot(data=titanic_df, x='Survived', y='Age', hue='Survived',
            palette='Blues', legend=False)
plt.xticks([0, 1], ['Did Not Survive', 'Survived'])
plt.xlabel('Survival Status')
plt.ylabel('Age')
plt.title('Age Distribution by Survival Status')

for i, survived in enumerate([0, 1]):
    ages = titanic_df[titanic_df['Survived'] == survived]['Age'].dropna()
    q1, median, q3 = ages.quantile([0.25, 0.5, 0.75])
    iqr = q3 - q1

    plt.text(i + 0.3, median + 1, f'Median: {median:.0f}', va='center', fontsize=8)
    plt.text(i + 0.3, q1 + 1, f'Q1: {q1:.0f}', va='center', fontsize=8)
    plt.text(i + 0.3, q3 + 1, f'Q3: {q3:.0f}', va='center', fontsize=8)
    plt.text(i + 0, q3 - 2, f'IQR: {iqr:.0f}', va='center', fontsize=8, color='navy')

plt.show()

```
