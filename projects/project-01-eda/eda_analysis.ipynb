```python
from datetime import datetime; 
print(f"Run: {datetime.now().strftime("%Y-%m-%d %H:%M:%S")}")

# Dataset Profile — row counts, field population, and value ranges

total_rows = len(df)
print(f"Total rows: {total_rows}\n")
print(f"{'Column':<15} {'Type':<10} {'Non-Null':>8} {'Null':>6} {'% Full':>8}  Range / Values")
print("-" * 80)

for col in df.columns:
    dtype = str(df[col].dtype)
    non_null = df[col].notna().sum()
    null = total_rows - non_null
    pct = non_null / total_rows * 100

    if df[col].dtype in ['float64', 'int64']:
        lo, hi = df[col].min(), df[col].max()
        detail = f"{lo:.2f} – {hi:.2f}"
    else:
        top = df[col].value_counts().head(3)
        detail = ", ".join([f"{v} ({c})" for v, c in top.items()])

    print(f"{col:<15} {dtype:<10} {non_null:>8} {null:>6} {pct:>7.1f}%  {detail}")
```

```python
from datetime import datetime; 
print(f"Run: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")

# Survival rate by age group
age_bins = [0, 12, 18, 35, 60, 80]
age_labels = ['Child (0-12)', 'Teen (13-18)', 'Young Adult (19-35)', 'Middle Age (36-60)', 'Senior (61+)']

df['AgeGroup'] = pd.cut(df['Age'], bins=age_bins, labels=age_labels)

survival_by_age = df.groupby('AgeGroup', observed=True)['Survived'].agg(['count', 'sum', 'mean'])
survival_by_age_by_class = df.groupby(['AgeGroup', 'Pclass'], observed=True)['Survived'].agg(['count', 'sum', 'mean'])
survival_by_age.columns = ['Total', 'Survived', 'SurvivalRate']
survival_by_age['SurvivalRate'] = survival_by_age['SurvivalRate'].map('{:.1%}'.format)

print("Survival Rate by Age Group")
print(f"Note: {df['Age'].isna().sum()} passengers excluded due to missing age\n")
display(survival_by_age)
display(survival_by_age_by_class)
```

```python
from datetime import datetime; 
print(f"Run: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")

# Survival rate by point of boarding
embarked_labels = {'S': 'Southampton', 'C': 'Cherbourg', 'Q': 'Queenstown'}
df['EmbarkedFull'] = df['Embarked'].map(embarked_labels)

survival_by_embarked = df.groupby('EmbarkedFull')['Survived'].agg(['count', 'sum', 'mean'])
survival_by_embarked.columns = ['Total', 'Survived', 'SurvivalRate']
survival_by_embarked['SurvivalRate'] = survival_by_embarked['SurvivalRate'].map('{:.1%}'.format)

print("Survival Rate by Point of Boarding")
print(f"Note: {df['Embarked'].isna().sum()} passengers excluded due to missing embarkation data\n")
display(survival_by_embarked)
```

```python
from datetime import datetime; 
print(f"Run: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")

# Survival rate by class by port of boarding
survival_by_class_by_port = df.groupby(['EmbarkedFull', 'Pclass'])['Survived'].agg(['count', 'sum', 'mean'])
survival_by_class_by_port.columns = ['Total', 'Survived', 'SurvivalRate']
survival_by_class_by_port['SurvivalRate'] = survival_by_class_by_port['SurvivalRate'].map('{:.1%}'.format)

print("Survival Rate by Class by Port of Boarding\n")
display(survival_by_class_by_port)
```

```python
from datetime import datetime; 
print(f"Run: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")

# Survival rate by sex and passenger class
survival_by_sex_class = df.groupby(['Sex', 'Pclass'])['Survived'].agg(['count', 'sum', 'mean'])
survival_by_sex_class.columns = ['Total', 'Survived', 'SurvivalRate']
survival_by_sex_class['SurvivalRate'] = survival_by_sex_class['SurvivalRate'].map('{:.1%}'.format)

print("Survival Rate by Sex and Passenger Class\n")
display(survival_by_sex_class)
```
